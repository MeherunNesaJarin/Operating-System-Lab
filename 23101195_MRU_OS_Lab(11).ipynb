```python
# Most Recently Used (MRU) Page Replacement

pages = [1, 2, 3, 4, 1, 2, 3, 1]
frame_size = 3
frames = []
page_fault = 0
page_hit = 0

for page in pages:
    if page in frames:
        page_hit += 1
        frames.remove(page)
        frames.append(page)
    else:
        page_fault += 1

        if len(frames) < frame_size:
            frames.append(page)
        else:
            # The last page is the most recently used
            mru_page = frames[-1]
            frames.remove(mru_page)
            frames.append(page)

    print(frames)

total_requests = len(pages)
hit_ratio = page_hit / total_requests

print("Page Faults:", page_fault)
print("Page Hits:", page_hit)
print("Page Hit Ratio:", hit_ratio)
```
