```python

class Process:
    def __init__(self, pid, at, bt, pr):
        self.pid = pid
        self.at = at
        self.bt = bt
        self.pr = pr
        self.ct = 0
        self.tat = 0
        self.wt = 0


# Input
n = int(input("Enter number of processes: "))
processes = []

for i in range(n):
    pid = "P" + str(i + 1)
    at = int(input("Arrival Time: "))
    bt = int(input("Burst Time: "))
    pr = int(input("Priority: "))

    processes.append(Process(pid, at, bt, pr))


# Priority Scheduling
time = 0
completed = 0

while completed < n:

    # Find available processes
    ready = []

    for p in processes:
        if p.at <= time and p.ct == 0:
            ready.append(p)

    # If no process is available
    if len(ready) == 0:
        time += 1
        continue

    # Select highest priority
    current = min(ready, key=lambda p: p.pr)

    # Execute process completely
    time += current.bt

    # Calculate values
    current.ct = time
    current.tat = current.ct - current.at
    current.wt = current.tat - current.bt

    completed += 1


# Output
print("\nPID\tAT\tBT\tPR\tCT\tTAT\tWT")

for p in processes:
    print(p.pid, "\t", p.at, "\t", p.bt, "\t",
          p.pr, "\t", p.ct, "\t", p.tat, "\t", p.wt)


# Average
avg_tat = sum(p.tat for p in processes) / n
avg_wt = sum(p.wt for p in processes) / n

print("\nAverage Turnaround Time:", round(avg_tat, 2))
print("Average Waiting Time:", round(avg_wt, 2))
```
