```python
#Non-preemptive SJF
processes = [
    ("P1", 3, 1),
    ("P2", 2, 4),
    ("P3", 3, 6),
    ("P4", 1, 3),
    ("P5", 4, 2),
    ("P6", 6, 1)
]

completed = []
ready_queue = []
time = 0
n = len(processes)

# Sort initially by arrival time
processes.sort(key=lambda x: x[1])

remaining_processes = processes.copy()

while len(completed) < n:
    # Add all arrived processes to ready_queue
    arrived = [p for p in remaining_processes if p[1] <= time]
    for p in arrived:
        ready_queue.append(p)
        remaining_processes.remove(p)

    if not ready_queue:
        time = remaining_processes[0][1]
        continue

    # Pick process with smallest BT
    ready_queue.sort(key=lambda x: x[2])
    p, at, bt = ready_queue.pop(0)

    time += bt
    ct = time
    tat = ct - at
    wt = tat - bt
    completed.append((p, at, bt, ct, tat, wt))

# Print Output
print("PID\tAT\tBT\tCT\tTAT\tWT")
total_tat, total_wt = 0, 0
for p, at, bt, ct, tat, wt in sorted(completed, key=lambda x: x[0]):
    print(f"{p}\t{at}\t{bt}\t{ct}\t{tat}\t{wt}")
    total_tat += tat
    total_wt += wt

print(f"\nAverage TAT = {total_tat/n:.2f}")
print(f"Average WT  = {total_wt/n:.2f}")
```
