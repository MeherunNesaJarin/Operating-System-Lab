```python
#SSTF Disk scheduling
head= int(input("Enter initial head: "))
requests = list(map(int,input("Enter requests: ").split()))

requests.sort()
total_seek=0

while requests:
  nearest = min(requests, key=lambda x: abs(head-x))
  distance = abs(nearest-head)
  print(head,"->",nearest,":",distance)
  total_seek += distance
  head = nearest
  requests.remove(nearest)
print("Total Head Movement=", total_seek)

```
