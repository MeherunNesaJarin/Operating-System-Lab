```python
#Preemptive SJF
processes = [
    {"p": "P1", "at": 3, "bt": 1, "rt": 1},
    {"p": "P2", "at": 2, "bt": 4, "rt": 4},
    {"p": "P3", "at": 3, "bt": 6, "rt": 6},
    {"p": "P4", "at": 1, "bt": 3, "rt": 3},
    {"p": "P5", "at": 4, "bt": 2, "rt": 2},
    {"p": "P6", "at": 6, "bt": 1, "rt": 1}
]

time = 0
completed = 0
n = len(processes)
total_tat, total_wt = 0, 0

print("PID\tAT\tBT\tCT\tTAT\tWT")

while completed < n:
    # Find process with minimum remaining time among arrived ones
    idx = -1
    min_rt = float('inf')

    for i in range(n):
        if processes[i]["at"] <= time and processes[i]["rt"] > 0:
            if processes[i]["rt"] < min_rt:
                min_rt = processes[i]["rt"]
                idx = i
            elif processes[i]["rt"] == min_rt:
                # Tie-breaker: earlier arrival time
                if processes[i]["at"] < processes[idx]["at"]:
                    idx = i

    if idx == -1:
        time += 1
        continue

    # Execute selected process for 1 unit
    processes[idx]["rt"] -= 1
    time += 1

    # Check if finished
    if processes[idx]["rt"] == 0:
        completed += 1
        ct = time
        at = processes[idx]["at"]
        bt = processes[idx]["bt"]
        tat = ct - at
        wt = tat - bt

        processes[idx]["ct"] = ct
        processes[idx]["tat"] = tat
        processes[idx]["wt"] = wt

        total_tat += tat
        total_wt += wt

# Sort output by process name
for p in sorted(processes, key=lambda x: x["p"]):
    print(f"{p['p']}\t{p['at']}\t{p['bt']}\t{p['ct']}\t{p['tat']}\t{p['wt']}")

print(f"\nAverage TAT = {total_tat/n:.2f}")
print(f"Average WT  = {total_wt/n:.2f}")
```
