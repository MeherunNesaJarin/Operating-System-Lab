```python
#Disk scheduling
head= int(input("Enter initial head position: "))

# Accept requests as a single, space-separated string
requests_str = input("Enter requests (space-separated): ")
requests = [int(x) for x in requests_str.split()]
n = len(requests) # 'n' now represents the number of requests entered

total_movement=0

print("\nFCFS Disk Scheduling")

for request in requests:
  distance = abs(head-request)
  print(f"{head}->{request} = {distance}")
  total_movement+=distance
  head=request

print(f"Total head movement =", total_movement)
```
