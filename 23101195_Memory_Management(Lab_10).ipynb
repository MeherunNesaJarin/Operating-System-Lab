**Frame Size = 3**

```python
#FIFO Page Replacement Algo
pages = [1, 2, 3, 4, 1, 2, 5, 1, 2, 3, 4, 5]
frame_size = 3
frames = []
page_fault = 0
page_hit = 0
for page in pages:
    # page hit
    if page in frames:
        page_hit += 1
        print(page, "-> HIT", frames)
    # page fault
    else:
        page_fault += 1
        if len(frames) < frame_size:
            frames.append(page)
        else:
            frames.pop(0)
            frames.append(page)
        print(page, "-> FAULT", frames)

print("\nTotal Page Fault =", page_fault)
print("Total Page Hit =", page_hit)

hit_ratio = page_hit / (page_hit + page_fault)
print("Hit Ratio =", hit_ratio)
```

**Frame Size = 4**

```python
#FIFO Page Replacement Algo
pages = [1, 2, 3, 4, 1, 2, 5, 1, 2, 3, 4, 5]
frame_size = 4

frames = []
page_fault = 0
page_hit = 0

for page in pages:
    # Page hit
    if page in frames:
        page_hit += 1
        print(page, "-> HIT", frames)

    # Page fault
    else:
        page_fault += 1

        if len(frames) < frame_size:
            frames.append(page)
        else:
            frames.pop(0)
            frames.append(page)

        print(page, "-> FAULT", frames)

print("\nTotal Page Fault =", page_fault)
print("Total Page Hit =", page_hit)

hit_ratio = page_hit / (page_hit + page_fault)
print("Hit Ratio =", hit_ratio)
```
