```python
# SRTF (Preemptive SJF) with Proper Gantt Chart
n = int(input("Enter number of processes: "))

process = []
for i in range(n):
    at = int(input(f"Enter Arrival Time of P{i+1}: "))
    bt = int(input(f"Enter Burst Time of P{i+1}: "))
    if bt <= 0:
        print("Burst time must be > 0")
        exit()
    process.append([i+1, at, bt, bt])   # [PID, AT, BT, Remaining BT]

time = 0
completed = 0
ct = [0] * n
gantt = [] # Will store tuples: (PID, start_time, end_time)

while completed < n:
    available = [p for p in process if p[1] <= time and p[3] > 0]

    if not available:
        time += 1
        continue

    # Shortest remaining time (Tie-breaker: FCFS)
    current = min(available, key=lambda x: (x[3], x[1]))
    pid = current[0]

    # Gantt Chart tracking
    if not gantt or gantt[-1][0] != pid:
        gantt.append([pid, time, time + 1])
    else:
        gantt[-1][2] = time + 1

    # Execute for 1 unit
    current[3] -= 1
    time += 1

    # Process completed
    if current[3] == 0:
        ct[pid - 1] = time
        completed += 1

# Calculate TAT and WT
print("\nProcess\tAT\tBT\tCT\tTAT\tWT")
total_tat = 0
total_wt = 0

for p in process:
    pid = p[0]
    at = p[1]
    bt = p[2]

    tat = ct[pid - 1] - at
    wt = tat - bt

    total_tat += tat
    total_wt += wt

    print(f"P{pid}\t{at}\t{bt}\t{ct[pid-1]}\t{tat}\t{wt}")

print(f"\nAverage TAT = {total_tat / n:.2f}")
print(f"Average WT  = {total_wt / n:.2f}")

print("\nGantt Chart:")
print("|", end="")
for pid, start, end in gantt:
    print(f"  P{pid}  |", end="")
print("\n0", end="")
for pid, start, end in gantt:
    print(f"{end:>6}", end="")
print()
```
