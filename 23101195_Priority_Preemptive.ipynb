```python
#For time quant 3s(priority_preemptive)
processes = [
    {"PID": "P1", "AT": 3, "BT": 3, "Priority": 1},
    {"PID": "P2", "AT": 0, "BT": 4, "Priority": 3},
    {"PID": "P3", "AT": 2, "BT": 2, "Priority": 2},
    {"PID": "P4", "AT": 1, "BT": 3, "Priority": 1},
    {"PID": "P5", "AT": 2, "BT": 4, "Priority": 0}
]

time_quantum = 3

for p in processes:
    p["Remaining"] = p["BT"]
    p["CT"] = 0
    p["TAT"] = 0
    p["WT"] = 0

time = 0
completed = 0
n = len(processes)

while completed < n:

    ready = []

    for p in processes:
        if p["AT"] <= time and p["Remaining"] > 0:
            ready.append(p)

    if len(ready) == 0:
        time += 1
        continue

    ready.sort(key=lambda x: (x["Priority"], x["AT"]))

    current = ready[0]

    run_time = min(time_quantum, current["Remaining"])

    for i in range(run_time):
        current["Remaining"] -= 1
        time += 1

        if current["Remaining"] == 0:
            break

    if current["Remaining"] == 0:
        current["CT"] = time
        current["TAT"] = current["CT"] - current["AT"]
        current["WT"] = current["TAT"] - current["BT"]
        completed += 1

print("PID\tAT\tBT\tPriority\tCT\tTAT\tWT")

total_wt = 0
total_tat = 0

for p in processes:
    print(
        p["PID"], "\t",
        p["AT"], "\t",
        p["BT"], "\t",
        p["Priority"], "\t\t",
        p["CT"], "\t",
        p["TAT"], "\t",
        p["WT"]
    )

    total_wt += p["WT"]
    total_tat += p["TAT"]

print("\nAverage Waiting Time:", total_wt / n)
print("Average Turnaround Time:", total_tat / n)
```
